## We were able to identify that the data was collected in India, and included 4 cities located inside the same province. Now we are going to create a hit map in orther to see it directly in the map

```python
%pip install folium
```

```python
import folium
from folium.plugins import HeatMap
```

```python
heat_data = [[row['latitude'], row['longitude']] for index, row in df2.iterrows()]

map_india = folium.Map(location=[20.5937, 78.9629], zoom_start=5,
                      tiles="OpenStreetMap")

HeatMap(heat_data).add_to(map_india)

map_india
```
